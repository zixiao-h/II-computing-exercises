```python
import scipy.integrate
import numpy as np
import matplotlib.pyplot as plt
```

```python
def deriv(t, y, tau):
    dy0 = y[1]
    dy1 = -(2.0/tau) * np.sin(y[0])**2 * y[1]
    return [dy0, dy1]

solution = scipy.integrate.solve_ivp(
    fun=deriv,
    t_span=(0,20),
    y0=(0.0,10.0),
    args=(2.0,),
    t_eval=np.linspace(0,20,100),
)
x, y, dydx = solution.t, solution.y[0], solution.y[1]
```

```python
fig, ax1 = plt.subplots()

ax1.plot(x, dydx, label="angular speed (rad/s)")
ax1.plot(x, 10*np.exp(-x/2.0), label="approximation")
ax1.legend(loc="lower right", bbox_to_anchor=(0.95,0.1))
ax1.set_xlabel("Time/s")
ax1.set_ylabel("Angular speed")
ax1.set_title("Evolution of a spinning ring in a magnetic field")

ax2 = ax1.twinx()
ax2.plot(x, y, label="angle (rad)", color="red")
ax2.set_ylabel("Position")
ax2.legend(loc="upper right", bbox_to_anchor=(0.95,0.9))
```
